```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

Ranks_list=[]
name_list=[]
club_list=[]
games_list=[]
goal_list=[]
minutes_list=[]
minutespergoal_list=[]
```

```python
url='https://www.indiansuperleague.com/stats/461-138-goals-award-statistics'
r=requests.get(url)
soup=BeautifulSoup(r.text,'lxml')
box=soup.find('div',class_="waf-body")

name=box.find_all('a',class_='player-name')

for n in name:
    names=n.text
    name_list.append(names)

print(name_list)
len(name_list)

```

```python
box1=soup.find("div",class_="table-body")
rank=box1.find_all('div',class_="table-data table-data-position")

for l in rank:
    ranks=l.text
    Ranks_list.append(ranks)

print(Ranks_list)
len(Ranks_list)
```

```python
box2=soup.find('div',class_="table-right")

team=box2.find_all('div',class_="name full-name")

for t in team:
    teams=t.text
    club_list.append(teams)

print(club_list)
len(club_list)
```

```python
goalss=box2.find_all('div',class_="table-data table-data-highlight undefined")

for g in goalss:
    goal=g.text
    goal_list.append(goal)

print(goal_list)
len(goal_list)
```

```python
minutess=box2.find_all('div',class_="table-data table-data-minutes-played undefined")

for m in minutess:
    minute=m.text
    minutes_list.append(minute)

minutes_list1=minutes_list[0::2]
minutespergoal_list=minutes_list[1::2]

print(minutes_list1)
print(minutespergoal_list)
```

```python
df=pd.DataFrame({"RANK":Ranks_list,
                 "PLAYER NAME":name_list,
                 "TEAM NAME":club_list,
                 "GOAL":goal_list,
                 "MINUTES":minutes_list1,
                 "MINUTES PER GOAL":minutespergoal_list})

print(df)
```
