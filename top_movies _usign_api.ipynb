```python
rows = []
for i in range(1, 500):
    Response = requests.get('https://api.themoviedb.org/3/discover/movie?api_key=fb736be990fb35750fe4b34a7451ae57&page={}'.format(i))
    temp_df=pd.DataFrame(Response.json()['results'])[['adult','id','title','popularity','release_date','vote_average','vote_count','original_language']]
    rows.append(temp_df)
    df = pd.concat(rows, ignore_index=True)
```
