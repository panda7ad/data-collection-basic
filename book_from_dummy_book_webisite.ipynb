```python
from bs4 import BeautifulSoup
```

```python
webpage=requests.get('https://books.toscrape.com/catalogue/page-1.html').text
```

```python
soup=BeautifulSoup(webpage , 'lxml')
```

```python
print(soup.prettify())
```

```python
price_list=[]
for i in soup.find_all('p',class_='price_color'):
    print(i.text.strip())
    price_list.append(i.text.strip())
```

```python
print(price_list)
```

```python
name_list=[]
for i in soup.find_all('h3'):
    print(i.text.strip())
    name_list.append(i.text.strip())
    
```

```python
print(name_list)
```

```python
stock_list=[]
for i in soup.find_all('p',class_='instock availability'):
    print(i.text.strip())
    stock_list.append(i.text.strip())
```

```python
prices = []
names = []
stocks = []

for y in range(1, 51):   # runs 1 → 50
    url = f'https://books.toscrape.com/catalogue/page-{y}.html'
    r = requests.get(url)

    # stop if no more pages
    if r.status_code != 200:
        print(f"Stopped at page {y} (no page found)")
        break

    soup = BeautifulSoup(r.text, 'lxml')
    print(f"\n--- Page {y} ---")

    for p in soup.find_all('p', class_='price_color'):
        prices.append(p.text.strip())

    for h in soup.find_all('h3'):
        names.append(h.text.strip())

    for s in soup.find_all('p', class_='instock availability'):
        stocks.append(s.text.strip())

print("\nScraping complete.")
print(f"Total books scraped: {len(names)}")
```

```python
d={'BOOK NAME':names,'PRICE':prices,'stock':stocks}
```

```python
df=pd.DataFrame(d)
```

```python
print(df)
```

```python
df.to_csv('1000 books.csv')
```
